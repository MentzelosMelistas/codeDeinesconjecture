```python
# Search all elliptic curves over Q with conductor <= 37
# Print their minimal discriminants and find the largest one.

db = CremonaDatabase()

largest_abs_D = 0
largest_D = None
largest_curve = None

for N in range(1, 38):

    for E in db.iter([N]):

        D = E.discriminant()

        print("Conductor:", N,
              " Label:", E.cremona_label(),
              " Minimal discriminant:", D)

        # Find the discriminant with largest absolute value
        if abs(D) > largest_abs_D:
            largest_abs_D = abs(D)
            largest_D = D
            largest_curve = E

print("\n========================================")
print("Largest absolute minimal discriminant:")
print("Curve:", largest_curve.cremona_label())
print("Conductor:", largest_curve.conductor())
print("Minimal discriminant:", largest_D)
print("Absolute value:", largest_abs_D)
print("========================================")
print (factor(largest_abs_D))
```
